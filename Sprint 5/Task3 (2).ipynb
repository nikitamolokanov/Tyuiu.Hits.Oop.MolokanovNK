```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void DisplayDelegate();
public delegate void AnimalHandler(string message);

public class Animal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }
    public string Habitat { get; set; }

    public event AnimalHandler AnimalUpdated;

    public Animal(string name, string species, int age, string habitat)
    {
        Name = name;
        Species = species;
        Age = age;
        Habitat = habitat;
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal animal)
        {
            return Name == animal.Name && Species == animal.Species;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Species);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Вид: {Species}, Возраст: {Age}, Среда: {Habitat}");
    }

    public void UpdateName(string newName)
    {
        Name = newName;
        AnimalUpdated?.Invoke($"Имя обновлено на: {newName}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public string Sound { get; set; }

    public Dog(string name, string species, int age, string habitat, string breed, string sound) 
        : base(name, species, age, habitat)
    {
        Breed = breed;
        Sound = sound;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Звук: {Sound}");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }

    public Cat(string name, string species, int age, string habitat, string color, bool isIndoor) 
        : base(name, species, age, habitat)
    {
        Color = color;
        IsIndoor = isIndoor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}, Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public Bird(string name, string species, int age, string habitat, double wingspan, bool canFly) 
        : base(name, species, age, habitat)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см, Умеет летать: {CanFly}");
    }
}

Dog dog = new Dog("Барон", "Млекопитающее", 5, "Двор", "Немецкая овчарка", "Гав");
Cat cat = new Cat("Мурка", "Млекопитающее", 3, "Дом", "Черный", true);
Bird bird = new Bird("Кеша", "Птица", 1, "Клетка", 25.5, true);

DisplayDelegate displayDog = dog.DisplayInfo;
DisplayDelegate displayCat = cat.DisplayInfo;
DisplayDelegate displayBird = bird.DisplayInfo;

displayDog();
displayCat();
displayBird();

List<Animal> animalList = new List<Animal> { dog, cat, bird };
var dogs = animalList.OfType<Dog>().Where(d => d.Age > 4);

foreach (var d in dogs)
{
    d.DisplayInfo();
}

Dictionary<string, Animal> animalDict = new Dictionary<string, Animal>
{
    { "dog1", dog },
    { "cat1", cat }
};

if (animalDict.TryGetValue("cat1", out Animal foundCat))
{
    foundCat.DisplayInfo();
}

HashSet<Animal> animalSet = new HashSet<Animal>();
animalSet.Add(dog);
animalSet.Add(cat);
animalSet.Add(bird);

var duplicateDog = new Dog("Барон", "Млекопитающее", 5, "Двор", "Немецкая овчарка", "Гав");
bool isAdded = animalSet.Add(duplicateDog);

AnimalHandler handler = (message) => Console.WriteLine(message);
dog.AnimalUpdated += handler;
dog.UpdateName("Шарик");
```
