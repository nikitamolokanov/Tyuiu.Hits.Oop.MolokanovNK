```
using System;
using System.Collections.Generic;

public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Habitat { get; set; }

    public Animal(string name, int age, string habitat)
    {
        Name = name;
        Age = age;
        Habitat = habitat;
    }

    public virtual void MakeSound(int volume)
    {
        Console.WriteLine($"Животное {Name} издает звук громкостью {volume} дБ. Среда обитания: {Habitat}.");
    }

    public void MakeSound()
    {
        Console.WriteLine($"Животное {Name} просто издает звук.");
    }

    public void MakeSound(string mood)
    {
        Console.WriteLine($"Животное {Name} издает звук в настроении: {mood}.");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public string FurColor { get; set; }
    public double Weight { get; set; }

    public Dog(string name, int age, string habitat, string breed, string furColor, double weight) 
        : base(name, age, habitat)
    {
        Breed = breed;
        FurColor = furColor;
        Weight = weight;
    }

    public override void MakeSound(int volume)
    {
        base.MakeSound(volume);
        Console.WriteLine($"Собака породы {Breed} (цвет: {FurColor}, вес: {Weight} кг) громко лает!");
    }

    public void Fetch()
    {
        Console.WriteLine($"Собака {Name} приносит мяч.");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }

    public Cat(string name, int age, string habitat, string color, bool isIndoor, int livesLeft) 
        : base(name, age, habitat)
    {
        Color = color;
        IsIndoor = isIndoor;
        LivesLeft = livesLeft;
    }

    public override void MakeSound(int volume)
    {
        base.MakeSound(volume);
        Console.WriteLine($"Кошка окраса {Color} (домашняя: {IsIndoor}, жизней: {LivesLeft}) мяукает!");
    }

    public void Scratch()
    {
        Console.WriteLine($"Кошка {Name} точит когти.");
    }
}

public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    public string FeatherColor { get; set; }

    public Bird(string name, int age, string habitat, double wingspan, bool canFly, string featherColor) 
        : base(name, age, habitat)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        FeatherColor = featherColor;
    }

    public override void MakeSound(int volume)
    {
        base.MakeSound(volume);
        Console.WriteLine($"Птица с размахом крыльев {Wingspan} см (летает: {CanFly}, окрас: {FeatherColor}) поет!");
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"Птица {Name} летит");
        else
            Console.WriteLine($"Птица {Name} не умеет летать.");
    }
}

public class AnimalCollection<T> where T : Animal
{
    private List<T> _animals = new List<T>();

    public void Add(T animal)
    {
        _animals.Add(animal);
    }

    public void Remove(T animal)
    {
        _animals.Remove(animal);
    }

    public void DisplayAnimals()
    {
        foreach (var animal in _animals)
        {
            animal.MakeSound(50);
        }
    }
}

Animal baseAnimal = new Animal("Неизвестное", 5, "Лес");
baseAnimal.MakeSound();
baseAnimal.MakeSound(30);
baseAnimal.MakeSound("Спокойное");
Console.WriteLine();

Dog myDog = new Dog("Барон", 3, "Двор", "Овчарка", "Черный", 30.5);
myDog.MakeSound(85);
myDog.Fetch();
Console.WriteLine();

Cat myCat = new Cat("Мурка", 2, "Дом", "Белый", true, 9);
myCat.MakeSound(40);
myCat.Scratch();
Console.WriteLine();

Bird myBird = new Bird("Кеша", 1, "Тропики", 25.5, true, "Зеленый");
myBird.MakeSound(15);
myBird.Fly();
Console.WriteLine();

AnimalCollection<Animal> collection = new AnimalCollection<Animal>();
collection.Add(myDog);
collection.Add(myCat);
collection.Add(myBird);
collection.DisplayAnimals();
```
