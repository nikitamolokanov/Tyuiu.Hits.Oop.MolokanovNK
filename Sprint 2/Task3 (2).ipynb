```
using System;

public abstract class Supplier
{
    public int SupplierId { get; set; }
    public string CompanyName { get; set; }
    public string ProductType { get; set; }
    public string ContactEmail { get; set; }

    protected Supplier(int supplierId, string companyName, string productType, string contactEmail)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        ContactEmail = contactEmail;
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID поставщика: {SupplierId}, Компания: {CompanyName}, Продукция: {ProductType}, Email: {ContactEmail}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine($"Компания {CompanyName} предоставляет котировку на товары типа: {ProductType}");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}.");
    }
}

public interface IExportable
{
    void ExportInfo();
}

public class Manufacturer : Supplier
{
    public int FoundedYear { get; set; }

    public Manufacturer(int supplierId, string companyName, string productType, string contactEmail, int foundedYear)
        : base(supplierId, companyName, productType, contactEmail)
    {
        FoundedYear = foundedYear;
    }

    public override void ProvideQuote()
    {
        base.ProvideQuote();
        Console.WriteLine($"Год основания компании: {FoundedYear}");
    }
}

public class Retailer : Supplier
{
    public string StoreLocation { get; set; }

    public Retailer(int supplierId, string companyName, string productType, string contactEmail, string storeLocation)
        : base(supplierId, companyName, productType, contactEmail)
    {
        StoreLocation = storeLocation;
    }

    public override void SubmitOrder()
    {
        base.SubmitOrder();
        Console.WriteLine($"Расположение магазина: {StoreLocation}");
    }
}

public class Importer : Supplier, IExportable
{
    public string OriginCountry { get; set; }
    public double ImportTaxRate { get; set; }

    public Importer(int supplierId, string companyName, string productType, string contactEmail, string originCountry, double importTaxRate)
        : base(supplierId, companyName, productType, contactEmail)
    {
        OriginCountry = originCountry;
        ImportTaxRate = importTaxRate;
    }

    public override void GetCompanyInfo()
    {
        base.GetCompanyInfo();
        Console.WriteLine($"Страна происхождения товара: {OriginCountry}");
    }

    public void ExportInfo()
    {
        Console.WriteLine($"Экспортные данные: Страна {OriginCountry}, Налоговая ставка {ImportTaxRate}%");
    }
}

public class InternationalManufacturer : Manufacturer
{
    public string InternationalLicense { get; set; }

    public InternationalManufacturer(int supplierId, string companyName, string productType, string contactEmail, int foundedYear, string internationalLicense)
        : base(supplierId, companyName, productType, contactEmail, foundedYear)
    {
        InternationalLicense = internationalLicense;
    }

    public override void ProvideQuote()
    {
        base.ProvideQuote();
        Console.WriteLine($"Международная лицензия: {InternationalLicense}");
    }
}

Manufacturer myManufacturer = new Manufacturer(101, "ООО ТехноПром", "Электроника", "tech@prom.ru", 2005);
myManufacturer.GetCompanyInfo();
myManufacturer.ProvideQuote();
myManufacturer.SubmitOrder();
Console.WriteLine();

Retailer myRetailer = new Retailer(102, "ИП Иванов", "Бытовая техника", "ivanov@shop.ru", "г. Москва, ул. Ленина, д. 10");
myRetailer.GetCompanyInfo();
myRetailer.ProvideQuote();
myRetailer.SubmitOrder();
Console.WriteLine();

Importer myImporter = new Importer(103, "Global Trade", "Одежда", "info@global.ru", "Китай", 15.5);
myImporter.GetCompanyInfo();
myImporter.ProvideQuote();
myImporter.SubmitOrder();
myImporter.ExportInfo();
Console.WriteLine();

InternationalManufacturer myIntManufacturer = new InternationalManufacturer(104, "MegaCorp", "Автозапчасти", "mega@corp.com", 1998, "ISO-9001");
myIntManufacturer.GetCompanyInfo();
myIntManufacturer.ProvideQuote();
myIntManufacturer.SubmitOrder();
```
