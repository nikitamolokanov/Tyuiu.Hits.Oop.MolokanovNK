```
using System;

public interface IOrderProcessor
{
    void ProcessOrder();
    string GetDeliveryDetails();
}

public abstract class Supplier
{
    public int SupplierId { get; set; }
    public string CompanyName { get; set; }
    public string ProductType { get; set; }
    public string ContactEmail { get; set; }

    protected Supplier(int supplierId, string companyName, string productType, string contactEmail)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        ContactEmail = contactEmail;
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID: {SupplierId}, Компания: {CompanyName}, Продукция: {ProductType}, Email: {ContactEmail}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine($"Компания {CompanyName} предоставляет котировку на {ProductType}");
    }
}

public class Manufacturer : Supplier, IOrderProcessor
{
    public int FoundedYear { get; set; }
    public string FactoryAddress { get; set; }

    public Manufacturer(int supplierId, string companyName, string productType, string contactEmail, int foundedYear, string factoryAddress) 
        : base(supplierId, companyName, productType, contactEmail)
    {
        FoundedYear = foundedYear;
        FactoryAddress = factoryAddress;
    }

    public override void ProvideQuote()
    {
        base.ProvideQuote();
        Console.WriteLine($"Год основания завода: {FoundedYear}");
    }

    public void ProcessOrder()
    {
        Console.WriteLine($"Заказ передан на производство {CompanyName}");
    }

    public string GetDeliveryDetails()
    {
        return $"Отгрузка со склада: {FactoryAddress}";
    }

    void IOrderProcessor.ProcessOrder()
    {
        Console.WriteLine($"Внутренняя обработка заказа на фабрике {CompanyName}");
    }
}

public class Retailer : Supplier, IOrderProcessor
{
    public string StoreLocation { get; set; }
    public int StoreSize { get; set; }

    public Retailer(int supplierId, string companyName, string productType, string contactEmail, string storeLocation, int storeSize) 
        : base(supplierId, companyName, productType, contactEmail)
    {
        StoreLocation = storeLocation;
        StoreSize = storeSize;
    }

    public void ProcessOrder()
    {
        Console.WriteLine($"Заказ принят в магазине {CompanyName}");
    }

    public string GetDeliveryDetails()
    {
        return $"Доставка в магазин по адресу: {StoreLocation}";
    }

    void IOrderProcessor.ProcessOrder()
    {
        Console.WriteLine($"Проверка наличия товара в магазине {StoreLocation}");
    }
}

public class OrderService
{
    private readonly IOrderProcessor _orderProcessor;

    public OrderService(IOrderProcessor orderProcessor)
    {
        _orderProcessor = orderProcessor;
    }

    public void ExecuteOrder()
    {
        _orderProcessor.ProcessOrder();
        Console.WriteLine(_orderProcessor.GetDeliveryDetails());
    }
}

Manufacturer myManufacturer = new Manufacturer(101, "ООО ТехноПром", "Электроника", "tech@prom.ru", 2005, "г. Москва, ул. Заводская, д. 5");
Retailer myRetailer = new Retailer(102, "ИП Иванов", "Бытовая техника", "ivanov@shop.ru", "г. Москва, ул. Ленина, д. 10", 150);

myManufacturer.GetCompanyInfo();
myManufacturer.ProvideQuote();
Console.WriteLine();

myRetailer.GetCompanyInfo();
myRetailer.ProvideQuote();
Console.WriteLine();

OrderService manufacturerService = new OrderService(myManufacturer);
manufacturerService.ExecuteOrder();
Console.WriteLine();

OrderService retailerService = new OrderService(myRetailer);
retailerService.ExecuteOrder();
Console.WriteLine();

((IOrderProcessor)myManufacturer).ProcessOrder();
((IOrderProcessor)myRetailer).ProcessOrder();
```
